```python
#Exercise 5: Union of Sets
set_a = {1, 2, 3, 4} 
set_b = {3, 4, 5, 6}
set_a | set_b

#alternatively
result = set_a.union(set_b)
print(f"result of union is {result}")
```

```python
#Exercise 6: Intersection of Sets
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a & set_b
#alternatively
result = set_a.intersection(set_b)
print(f"result of intersection is {result}")
```

```python
#Exercise 7: Difference of Sets
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.difference(set_b)
#alternatively
result = set_a - set_b
print(f"result of difference is {result}")
```

```python
#Exercise 8: Symmetric Difference
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.symmetric_difference(set_b)
```

```python
#Exercise 14: Intersection Check with isdisjoint()
set_a = {1, 2, 3} 
set_b = {4, 5, 6}
set_a.isdisjoint(set_b)
```

```python
import pandas as pd

# Create a DataFrame
df = pd.DataFrame({
    'set_a': [1, 2, 3],
    'set_b': [4, 5, 6]
})

# Convert columns to sets
set_a = set(df['set_a'])
set_b = set(df['set_b'])

# Check whether the sets are disjoint
result = set_a.isdisjoint(set_b)

print("Are the sets disjoint?", result)
```
